The two datasets below explore the impact of climate change on agricultural productivity across various regions. They include data on temperature, precipitation, and crop yields to understand how shifting climate patterns affect agriculture.

### Task 15: 
* Evaluate the distribution of the "Crop_Yield_MT_per_HA" variable.
* Is the distribution normal or skewed?
* Should the variable be transformed?
* Why or why not?
* If the variable needs to be transformed, transform it using the square root method and store the transformed values in a new column called "Crop_Yield_MT_per_HA_transformed".
* Visualize the distribution before and after the transformation.
* Write a few sentences explaining why (or why not) you decided to transform or not to transform the data in the context of this dataset.
* Write your answer using markdown.



```python
# Task 15 should be completed in the cells below.  You can create additional cells if needed
```

```python
sns.histplot(crops_df['Crop_Yield_MT_per_HA'], bins=50, kde=True)
```

```python
sns.boxplot(data=crops_df['Crop_Yield_MT_per_HA'])
```

The histogram is slightly right skewed along with the boxplot even though there are not outliers. It is still a good idea to scale. 

```python
import math
crops_df['Crop_Yield_MT_per_HA_transformed'] = crops_df['Crop_Yield_MT_per_HA'].apply(lambda x: math.sqrt(x))
```

```python
crops_df
```

```python
sns.histplot(crops_df['Crop_Yield_MT_per_HA_transformed'], bins=50, kde=True)
```

```python
sns.boxplot(data=crops_df['Crop_Yield_MT_per_HA_transformed'])
```

Now it is evenly distributed.
