```python
def split_salary_col(val):

    #checkk for null & NaN
    if val == None or math.isnan(val):
        val = ""
    
    
    #convert value to lower case
    val = val.lower().replace("$","")
    data_source = ""
    
    #extract source part
    if val.find("(glassdoor est.)")!=-1:
        data_source = "Glassdoor"
    elif val.find("(employer set.)")!=-1:
        data_source = "Employer"
    else:
        data_source = "Uknown"
        
        #remove source part and replace k with 000
    val = str(val).replace("(glassdoor est.)","").replace("(employer set.)","").strip()
    val = val.replace("k","000")

    #split value into lower and upper bound
    temp = val.split("-")

    #check is only lower exists then duplicate it   
    if len(temp) == 1:
        temp.append(temp[0])

    #deal with blank/missing/null values
    for i in range(0,len(temp)):
        temp[i] = temp[i].strip().replace("","0")
        if temp[i]== "":
            temp[i]=0

    print("range: ",temp)
    print("val: ", val)
    return data_source, float(temp[0]),float(temp[1])
```
