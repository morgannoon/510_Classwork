## Concat

**Pandas Cheat sheet** https://pandas.pydata.org/Pandas_Cheat_Sheet.pdf

#### Concat by rows

**Assumptions:**
* Two or more dataframes
* Each dataframe has the same number of columns
* Columns have identical headers
* columns in all dataframes appear in the same order

```python
df1 = df.iloc[:3,:5]
df1.head()
```

```python
df2 = df.iloc[4:6,:5]
df2.head()
```

```python
df3 = pd.concat([df1,df2],axis =0) #same columns combine rows
df3
```

### Concat by Columns

#### Concat by columns

**Assumptions:**
* Two or more dataframes
* Each dataframe has the same number of rows
* Rows have identical indences
* Rows in all dataframes appear in the same order

```python
df1 = df.iloc[:3,[0,1,2]]
df1
```

```python
df2 = df.iloc[:3,[3,4,5]]
df2
```

```python
df3 = pd.concat([df1,df2],axis=1) #same rows combine columns
df3
```

## Merging DataFrame(s)

```python
df.head()
```

```python
df1 = df.iloc[:3,:3]
df1
```

```python
df2 = df.iloc[1:4,[0,4,5,6]]
df2
```

**Let's compare the two data frames**
* both have rows 1 & 2 
* both have a column called id
  
**In Pandas, instead of using SQL keyword "JOIN", we use a method called "merge()"**

```python
df3 = pd.merge(df1,df2,how='inner',left_on = "id", right_on= "id")
df3 ##only the two rows that were in both data frames will show (inner combine both)
```

**left join will show all rows in left table but will show missing data as NaN**

```python
df4 = pd.merge(df1,df2,how='left',left_on = "id", right_on= "id")
df4 ## all rows of left data frame and those in both data frames will show
```

**right join will show all rows in right table but will show missing data as NaN**

```python
df5 = pd.merge(df1,df2,how='right',left_on = "id", right_on= "id")
df5 # all rows of right data frame and those in both data frames will show
# equivalent to reversing fd1 and fd2 in a left merge
```

**full join (outer join) will show all rows in left and right data frame**

```python
df6 = pd.merge(df1,df2, how='outer',left_on='id',right_on='id')
df6 #will combine left and right and missing data will be NaN
```

## Querying

```python
df.head()
```

```python
q = "chol > 350" #queries pass as strings
df.query(q)
```

## Grouping

```python
df.head()
```

```python
df['frame'].unique()
```

```python
df['frame'].value_counts()
```

```python
df['gender'].value_counts()
```

```python
d = { "banana":1.99, "apple":5.99,"pear":7.99}
print(d)
```

```python
#{"chol":"mean","age":"median","weight":"max"} pass a dictionary
```

```python
#avg chol by gender
df.groupby('gender').agg({"chol":"mean"})
```

```python
#avg chol, age, weight by frame
df.groupby('frame').agg({"chol":"mean","age":"median","weight":"max"})
```

```python
df_frame= df.groupby('frame').agg({"chol":"mean","age":"median","weight":"max"})
df_frame.head()
```

```python
df_frame.reset_index() #default pandas non destructive will not override things
#only shows what it will look like does not change dataset
```

```python
df_frame.head()
```

```python
df_frame.reset_index(inplace = True) #this will make a change to df_frame(destructive)
```

```python
df_frame.head() 
```
